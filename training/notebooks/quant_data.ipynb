```python
from pathlib import Path

import numpy as np
from PIL import Image


data_root = Path(r"../../datasets/images/oscd96/test")
sample_id = "44"
out_path = Path(f"oscd_44.npz")

image_a = np.asarray(Image.open(data_root / "A" / f"{sample_id}.png"))
image_b = np.asarray(Image.open(data_root / "B" / f"{sample_id}.png"))
label = np.asarray(Image.open(data_root / "label" / f"{sample_id}.png"))

image_pair = np.concatenate([image_a, image_b], axis=-1).astype(np.float32)
image_pair = image_pair.transpose(2, 0, 1)[None]
label = label[None][None].astype(np.float32)#.transpose(1, 2, 0)

np.savez(out_path, x_test=image_pair, y_test=label)

print(f"saved {out_path}")
print("INPUT", image_pair.shape, image_pair.dtype, image_pair.min(), image_pair.max())
print("OUTPUT", label.shape, label.dtype, label.min(), label.max())
```
